```python
# NAME: rentalcar.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
def get_valid_input(prompt, valid_options):
    while True:
        user_input = input(prompt).strip().upper()
        if user_input in valid_options:
            return user_input
        else:
            print(f"Invalid input. Please enter one of {', '.join(valid_options)}.")

def get_numeric_input(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value >= 0:
                return value
            else:
                print("Input must be a non-negative number.")
        except ValueError:
            print("Invalid input. Please enter a number.")

def calculate_bill(car_type, rental_days, distance_driven):
    base_daily_rate = 0
    mileage_rate = 0

    if car_type == 'B':
        base_daily_rate = 40.00
        mileage_rate = 0.25
    elif car_type == 'S':
        base_daily_rate = 60.00
        mileage_rate = 0.35
    elif car_type == 'L':
        base_daily_rate = 90.00
        mileage_rate = 0.45

    total_cost = (base_daily_rate * rental_days) + (mileage_rate * distance_driven)
    return total_cost

# PROGRAM LOGIC
print("Welcome to the Rental Car System!")

# Collect user input for car type
car_type = get_valid_input("Enter car type (B for Budget, S for Standard, L for Luxury): ", ['B', 'S', 'L'])
print(f"You selected: {car_type}")

# Collect number of rental days
rental_days = get_numeric_input("Enter number of rental days: ")

# Collect odometer readings
start_odometer = get_numeric_input("Enter start odometer reading: ")
end_odometer = get_numeric_input("Enter end odometer reading: ")

# Ensure end odometer is not less than start odometer
while end_odometer < start_odometer:
    print("End odometer reading cannot be less than start odometer reading. Please re-enter.")
    end_odometer = get_numeric_input("Enter end odometer reading: ")

# Calculate distance driven
distance_driven = end_odometer - start_odometer
print(f"Distance driven: {distance_driven:.2f} miles")

# Calculate the final bill
final_bill = calculate_bill(car_type, rental_days, distance_driven)

# Print the final report
print("\n--- Rental Summary --- ")
print(f"Car Type: {car_type}")
print(f"Rental Days: {rental_days:.0f}")
print(f"Distance Driven: {distance_driven:.2f} miles")
print(f"Total Cost: ${final_bill:.2f}")
```
