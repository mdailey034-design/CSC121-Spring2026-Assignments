```python
# NAME: roster.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
def add_players(player_list):
    while True:
        player_name = input("Enter player name (type 'quit' to finish): ")
        if player_name.lower() == 'quit':
            break
        player_list.append(player_name)

def is_on_roster(player_list, name_to_find):
    return name_to_find in player_list

# PROGRAM LOGIC
roster = []
add_players(roster)
print("Current Roster:", roster)

search_name = input("Enter a name to check if they are on the roster: ")
if is_on_roster(roster, search_name):
    print(f"{search_name} is on the roster.")
else:
    print(f"{search_name} is not on the roster.")
```
