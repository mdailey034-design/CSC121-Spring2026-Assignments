```python
# NAME: journal.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
# Function 1 (write_entry): * This function should accept a filename and a string (the entry) as parameters.

# It must open the file in append mode ('a').

# It should write the entry to the file, ensuring each new entry starts on a new line.

# Don't forget to close the file when finished.
def write_entry(filename, entry):
    with open(filename, 'a') as f:
        f.write(entry + '\n')

# Function 2 (read_journal):

# This function should accept the filename as a parameter.

# It should open the file in read mode ('r').

# It should read the entire contents and print them to the screen so the user can see their past entries.
def read_journal(filename):
    try:
        with open(filename, 'r') as f:
            print("\n--- Your Journal Entries ---")
            print(f.read())
            print("--------------------------")
    except FileNotFoundError:
        print("No journal entries found yet.")

# PROGRAM LOGIC
# Ask the user if they want to (W)rite a new entry or (R)ead their journal.

# If they choose 'W', ask for their thoughts and call write_entry.

# If they choose 'R', call read_journal.

# Wrap this in a while loop so the user can keep using the app until they type "quit".

journal_filename = "journal.txt"

while True:
    choice = input("Do you want to (W)rite a new entry or (R)ead your journal? (Type 'quit' to exit): ").lower()

    if choice == 'w':
        entry = input("Enter your journal entry: ")
        write_entry(journal_filename, entry)
        print("Entry saved.")
    elif choice == 'r':
        read_journal(journal_filename)
    elif choice == 'quit':
        print("Exiting journal app. Goodbye!")
        break
    else:
        print("Invalid choice. Please enter 'W', 'R', or 'quit'.")
```
