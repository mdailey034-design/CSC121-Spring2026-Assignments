```python
def write_entry(filename, entry):
    with open(filename, 'a') as f:
        f.write(entry + '\n')

def read_journal(filename):
    try:
        with open(filename, 'r') as f:
            print("\n--- Your Journal Entries ---")
            print(f.read())
            print("--------------------------")
    except FileNotFoundError:
        print("No journal entries found yet.")

journal_filename = "journal.txt"

while True:
    choice = input("Do you want to (W)rite a new entry or (R)ead your journal? (Type 'quit' to exit): ").lower()

    if choice == 'w':
        entry = input("Enter your journal entry: ")
        write_entry(journal_filename, entry)
        print("Entry saved.")
    elif choice == 'r':
        read_journal(journal_filename)
    elif choice == 'quit':
        print("Exiting journal app. Goodbye!")
        break
    else:
        print("Invalid choice. Please enter 'W', 'R', or 'quit'.")
```
