```python
# NAME: gradebook.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
def add_students(gradebook):
    """
    Prompts the user to input student names and their grades,
    adding them to the provided gradebook dictionary.
    """
    while True:
        student_name = input("Enter student's name (or type 'done' to finish): ")
        if student_name.lower() == 'done':
            break

        while True:
            try:
                grade_str = input(f"Enter {student_name}'s grade (0-100): ")
                grade = int(grade_str)
                if 0 <= grade <= 100:
                    gradebook[student_name] = grade
                    break
                else:
                    print("Grade must be between 0 and 100. Please try again.")
            except ValueError:
                print("Invalid input. Please enter a number for the grade.")

def calculate_average(gradebook):
    """
    Calculates and returns the average grade of all students in the gradebook.
    """
    if not gradebook:
        return 0.0  # Handle division by zero if no students are entered
    total_grades = sum(gradebook.values())
    return total_grades / len(gradebook)
```

```python
# PROGRAM LOGIC
gradebook = {}
print("--- Adding Students ---")
add_students(gradebook)

print("--- Gradebook Contents ---")
if gradebook:
    for student, grade in gradebook.items():
        print(f"{student}: {grade}")
else:
    print("No students were added to the gradebook.")

print("--- Class Average ---")
average_grade = calculate_average(gradebook)
print(f"Class Average: {average_grade:.2f}")  # Format to two decimal places
```
