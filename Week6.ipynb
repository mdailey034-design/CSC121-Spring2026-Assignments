```python
import math

def area(radius):
    return math.pi * radius**2

def circumference(radius):
    return 2 * math.pi * radius

def volume(radius, height):
    return area(radius) * height

# PROGRAM LOGIC
try:
    radius = float(input("Enter the radius: "))
    if radius < 0:
        print("Radius cannot be negative.")
    else:
        calculated_area = area(radius)
        calculated_circumference = circumference(radius)
        print(f"Area of the circle: {calculated_area:.2f}")
        print(f"Circumference of the circle: {calculated_circumference:.2f}")

        # Corrected indentation and syntax for height input and check
        height = float(input("Enter the height: "))
        if height < 0:
            print("Height cannot be negative.")
        else:
            calculated_volume = volume(radius, height)
            print(f"Volume of the cylinder: {calculated_volume:.2f}")
except ValueError:
    print("Invalid input. Please enter a numerical value for radius and height.")
```
