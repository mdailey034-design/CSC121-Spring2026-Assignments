```python

# IMPORT STATEMENTS
# 1. Import math

# FUNCTION DEFINITIONS
# 1. Define area(radius): returns pi * r squared
# 2. Define circumference(radius): returns 2 * pi * r
# 3. Define volume(radius, height):
#    - Call area(radius) and multiply by height.
#    - Return the result


import math

def area(radius):
    return math.pi * radius**2

def circumference(radius):
    return 2 * math.pi * radius

def volume(radius, height):
    return area(radius) * height

# PROGRAM LOGIC
try:
    radius = float(input("Enter the radius: "))
    if radius < 0:
        print("Radius cannot be negative.")
    else:
        calculated_area = area(radius)
        calculated_circumference = circumference(radius)
        print(f"Area of the circle: {calculated_area:.2f}")
        print(f"Circumference of the circle: {calculated_circumference:.2f}")
```
