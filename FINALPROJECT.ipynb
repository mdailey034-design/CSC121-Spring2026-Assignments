```python
expenses = []

def add_expense(description, category, amount):
    expense = {
        'description': description,
        'category': category,
        'amount': amount
    }
    expenses.append(expense)
    print(f"Added: {description} ({category}) - ${amount:.2f}")
```

```python
def total_expenses():
    """Calculates the total of all expenses in the list."""
    total = sum(expense['amount'] for expense in expenses)
    return total

def category_total(category_name):
    """Calculates the total expenses for a specific category."""
    total = sum(expense['amount'] for expense in expenses if expense['category'].lower() == category_name.lower())
    return total

def description_total(keyword):
    """Calculates the total expenses for items matching a keyword in their description.
    Used as a 'vendor' total since a dedicated vendor field is not present."""
    total = sum(expense['amount'] for expense in expenses if keyword.lower() in expense['description'].lower())
    return total
```

```python
add_expense('Groceries', 'Food', 55.23)
add_expense('Dinner with friends', 'Food', 75.00)
add_expense('Electricity bill', 'Utilities', 120.50)
add_expense('New book', 'Entertainment', 15.99)
add_expense('Gasoline', 'Transportation', 40.00)
add_expense('More Groceries', 'Food', 30.10)

print("\nRaw list of expenses:")
for expense in expenses:
    print(expense)
```

```python
print("\n--- Expense Report ---")
print(f"Total Expenses: ${total_expenses():.2f}")

print("\n--- Category Totals ---")
print(f"Total Food Expenses: ${category_total('Food'):.2f}")
print(f"Total Utilities Expenses: ${category_total('Utilities'):.2f}")
print(f"Total Entertainment Expenses: ${category_total('Entertainment'):.2f}")
print(f"Total Transportation Expenses: ${category_total('Transportation'):.2f}")

print("\n--- Description (Vendor) Totals ---")
print(f"Total 'Groceries' related Expenses: ${description_total('Groceries'):.2f}")
print(f"Total 'bill' related Expenses: ${description_total('bill'):.2f}")
```
