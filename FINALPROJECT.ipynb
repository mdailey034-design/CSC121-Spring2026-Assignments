```python
expenses = []

def add_expense(description, category, amount):
    expense = {
        'description': description,
        'category': category,
        'amount': amount
    }
    expenses.append(expense)
    print(f"Added: {description} ({category}) - ${amount:.2f}")
```

### Testing the Expense Input

Let's add a few sample expenses to our list and then print the raw list to ensure the data is being stored correctly.

```python
add_expense('Groceries', 'Food', 55.23)
add_expense('Dinner with friends', 'Food', 75.00)
add_expense('Electricity bill', 'Utilities', 120.50)
add_expense('New book', 'Entertainment', 15.99)
add_expense('Gasoline', 'Transportation', 40.00)
add_expense('More Groceries', 'Food', 30.10)

print("\nRaw list of expenses:")
for expense in expenses:
    print(expense)
```
