```python
# NAME: rectangle.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
# This function should accept two parameters: length and width.
def calculate_rectangle(length, width):
    # It should calculate the Area (length * width) and the Perimeter (2 * (length + width)).
    area = length * width
    perimeter = 2 * (length + width)
    # It must return both results as a single tuple in the format: (area, perimeter).
    return (area, perimeter)

# PROGRAM LOGIC
# Ask the user to input a length and a width (convert these to floats).
user_length = float(input("Enter the length of the rectangle: "))
user_width = float(input("Enter the width of the rectangle: "))

# Call your calculate_rectangle function using the user's input.
user_area, user_perimeter = calculate_rectangle(user_length, user_width)

# Unpack the returned tuple into two separate variables called my_area and my_perimeter.

# Print the results clearly for the user.
print(f"\nFor user input length = {user_length} and width = {user_width}:")
print(f"The calculated Area is: {user_area}")
print(f"The calculated Perimeter is: {user_perimeter}")

# Test the function with fixed numbers
test_length = 5
test_width = 10
test_area, test_perimeter = calculate_rectangle(test_length, test_width)
print(f"\nFor length = {test_length} and width = {test_width}:")
print(f"Area = {test_area}")
print(f"Perimeter = {test_perimeter}")
```
