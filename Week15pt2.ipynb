# Python Video Game Collection Manager

This program will allow you to manage your video game collection. It includes functionality to add games, view your collection, filter by various criteria, and save/load your data to a JSON file.

```python
import json

class Game:
    def __init__(self, title, platform, genre, rating, completion_status):
        self.title = title
        self.platform = platform
        self.genre = genre
        self.rating = rating
        self.completion_status = completion_status

    def to_dict(self):
        return {
            'title': self.title,
            'platform': self.platform,
            'genre': self.genre,
            'rating': self.rating,
            'completion_status': self.completion_status
        }

    @classmethod
    def from_dict(cls, data):
        return cls(data['title'], data['platform'], data['genre'], data['rating'], data['completion_status'])

    def __str__(self):
        return f"Title: {self.title}, Platform: {self.platform}, Genre: {self.genre}, Rating: {self.rating}/10, Status: {self.completion_status}"
```

```python
GAME_FILE = 'game_collection.json'
game_collection = []

def load_games():
    global game_collection
    try:
        with open(GAME_FILE, 'r') as f:
            data = json.load(f)
            game_collection = [Game.from_dict(d) for d in data]
        print(f"Loaded {len(game_collection)} games from {GAME_FILE}")
    except FileNotFoundError:
        print("No existing game collection found. Starting fresh.")
    except json.JSONDecodeError:
        print("Error reading game file. Starting fresh.")
    except Exception as e:
        print(f"An unexpected error occurred while loading games: {e}")

def save_games():
    try:
        with open(GAME_FILE, 'w') as f:
            json.dump([game.to_dict() for game in game_collection], f, indent=4)
        print(f"Saved {len(game_collection)} games to {GAME_FILE}")
    except IOError as e:
        print(f"Error saving games to file: {e}")
    except Exception as e:
        print(f"An unexpected error occurred while saving games: {e}")
```

```python
def add_game():
    while True:
        title = input("Enter game title: ").strip()
        if title:
            break
        else:
            print("Game title cannot be empty. Please try again.")

    while True:
        platform = input("Enter platform (e.g., PC, PlayStation, Xbox): ").strip().title()
        if platform:
            break
        else:
            print("Platform cannot be empty. Please try again.")

    while True:
        genre = input("Enter genre (e.g., RPG, Action, Strategy): ").strip().title()
        if genre:
            break
        else:
            print("Genre cannot be empty. Please try again.")

    while True:
        try:
            rating = int(input("Enter your rating (1-10): "))
            if 1 <= rating <= 10:
                break
            else:
                print("Rating must be between 1 and 10.")
        except ValueError:
            print("Invalid input. Please enter a number for rating.")

    while True:
        completion_status = input("Enter completion status (e.g., Completed, Playing, Backlog): ").strip().title()
        if completion_status:
            break
        else:
            print("Completion status cannot be empty. Please try again.")

    new_game = Game(title, platform, genre, rating, completion_status)
    game_collection.append(new_game)
    print(f"'{title}' added to your collection!")
```

```python
def view_games(games_to_display=None):
    if games_to_display is None:
        games_to_display = game_collection

    if not games_to_display:
        print("Your collection is empty.")
        return

    print("\n--- Your Game Collection ---")
    for i, game in enumerate(games_to_display):
        print(f"{i+1}. {game}")
    print("--------------------------")
```

```python
def filter_games():
    print("\nFilter options:")
    print("1. Filter by platform")
    print("2. Filter by completion status")
    print("3. Back to main menu")

    choice = input("Enter your choice: ")

    filtered_games = []
    if choice == '1':
        platform_filter = input("Enter platform to filter by: ").strip()
        filtered_games = [game for game in game_collection if game.platform.lower() == platform_filter.lower()]
        view_games(filtered_games)
    elif choice == '2':
        status_filter = input("Enter completion status to filter by: ").strip()
        filtered_games = [game for game in game_collection if game.completion_status.lower() == status_filter.lower()]
        view_games(filtered_games)
    elif choice == '3':
        return
    else:
        print("Invalid choice. Please try again.")
```

```python
def analyze_collection():
    print("\nAnalysis options:")
    print("1. Average rating by genre")
    print("2. Average rating by platform")
    print("3. Back to main menu")

    choice = input("Enter your choice: ")

    if choice == '1':
        genre_ratings = {}
        for game in game_collection:
            genre_ratings.setdefault(game.genre, []).append(game.rating)

        if not genre_ratings:
            print("No games in collection to analyze by genre.")
            return

        print("\n--- Average Rating by Genre ---")
        for genre, ratings in genre_ratings.items():
            avg_rating = sum(ratings) / len(ratings)
            print(f"{genre}: {avg_rating:.2f}/10")
        print("------------------------------")
    elif choice == '2':
        platform_ratings = {}
        for game in game_collection:
            platform_ratings.setdefault(game.platform, []).append(game.rating)

        if not platform_ratings:
            print("No games in collection to analyze by platform.")
            return

        print("\n--- Average Rating by Platform ---")
        for platform, ratings in platform_ratings.items():
            avg_rating = sum(ratings) / len(ratings)
            print(f"{platform}: {avg_rating:.2f}/10")
        print("--------------------------------")
    elif choice == '3':
        return
    else:
        print("Invalid choice. Please try again.")
```

```python
def main_menu():load_games()
while True:
        print("\n--- Game Collection Manager ---")
        print("1. Add a new game")
        print("2. View all games")
        print("3. Filter games")
        print("4. Analyze collection")
        print("5. Save and Exit")
        print("------------------------------")

        choice = input("Enter your choice: ")

        if choice == '1':
            add_game()
        elif choice == '2':
            view_games()
        elif choice == '3':
            filter_games()
        elif choice == '4':
            analyze_collection()
        elif choice == '5':
            save_games()
            print("Exiting Game Collection Manager. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")

# To run the program, simply call main_menu()
main_menu()
```
