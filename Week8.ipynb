```python
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
def is_palindrome(word):
  """
  Checks if a given word is a palindrome (reads the same forwards and backwards).

  Args:
    word: The string to be checked.

  Returns:
    True if the word is a palindrome, False otherwise.
  """
  # Convert the word to lowercase to handle case-insensitivity
  cleaned_word = str(word).lower()
  # Compare the cleaned word with its reverse
  return cleaned_word == cleaned_word[::-1]

# Example usage:
print(f"'madam' is a palindrome: {is_palindrome('madam')}")
print(f"'racecar' is a palindrome: {is_palindrome('racecar')}")
print(f"'hello' is a palindrome: {is_palindrome('hello')}")
print(f"'level' is a palindrome: {is_palindrome('level')}")
print(f"'AmoMA' is a palindrome: {is_palindrome('AmoMA')}")

# PROGRAM LOGIC
```
