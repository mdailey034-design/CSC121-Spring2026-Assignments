### Personal Task Manager

This script provides a simple command-line interface for managing personal tasks. You can add new tasks with a name and priority level, view all current tasks, and exit the program.

```python
# Initialize an empty list to store tasks
tasks = []

def add_task():
    """Adds a new task with a name and priority level."""
    task_name = input("Enter task name: ").strip()
    if not task_name:
        print("Task name cannot be empty. Please try again.")
        return

    while True:
        try:
            priority_input = input("Enter priority (1 for High, 2 for Medium, 3 for Low): ")
            priority = int(priority_input)
            if priority not in [1, 2, 3]:
                raise ValueError
            break
        except ValueError:
            print("Invalid priority. Please enter 1, 2, or 3.")

    task = {'name': task_name, 'priority': priority}
    tasks.append(task)
    print(f"Task '{task_name}' added with priority {priority}.")

def view_tasks():
    """Displays all tasks, sorted by priority."""
    if not tasks:
        print("No tasks to display.")
        return

    # Sort tasks by priority (1=High, 2=Medium, 3=Low)
    sorted_tasks = sorted(tasks, key=lambda x: x['priority'])

    print("\n--- Your Tasks ---")
    for i, task in enumerate(sorted_tasks):
        priority_map = {1: 'High', 2: 'Medium', 3: 'Low'}
        print(f"{i+1}. Name: {task['name']}, Priority: {priority_map[task['priority']]}")
    print("------------------\n")

def main():
    """Main function to run the task manager."""
    while True:
        print("\nPersonal Task Manager Menu:")
        print("1. Add a task")
        print("2. View tasks")
        print("3. Exit")

        choice = input("Enter your choice (1-3): ")

        if choice == '1':
            add_task()
        elif choice == '2':
            view_tasks()
        elif choice == '3':
            print("Exiting Task Manager. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter 1, 2, or 3.")

# Run the main program
if __name__ == "__main__":
    main()
```
