### Personal Task Manager

This script provides a simple command-line interface for managing personal tasks. You can add new tasks with a name and priority level, view all current tasks, and exit the program.

```python
import datetime # Import datetime for timestamping completed tasks

# Initialize an empty list to store tasks
tasks = []

def add_task():
    """Adds a new task with a name and priority level."""
    task_name = input("Enter task name: ").strip()
    if not task_name:
        print("Task name cannot be empty. Please try again.")
        return

    while True:
        try:
            priority_input = input("Enter priority (1 for High, 2 for Medium, 3 for Low): ")
            priority = int(priority_input)
            if priority not in [1, 2, 3]:
                raise ValueError
            break
        except ValueError:
            print("Invalid priority. Please enter 1, 2, or 3.")

    task = {'name': task_name, 'priority': priority}
    tasks.append(task)
    print(f"Task '{task_name}' added with priority {priority}.")

def view_tasks():
    """Displays all tasks, sorted by priority."""
    if not tasks:
        print("No tasks to display.")
        return

    # Sort tasks by priority (1=High, 2=Medium, 3=Low)
    sorted_tasks = sorted(tasks, key=lambda x: x['priority'])

    print("\n--- Your Tasks ---")
    for i, task in enumerate(sorted_tasks):
        priority_map = {1: 'High', 2: 'Medium', 3: 'Low'}
        print(f"{i+1}. Name: {task['name']}, Priority: {priority_map[task['priority']]}")
    print("------------------\n")

def complete_task():
    """Marks a task as complete, removes it from the list, and saves its name to a file."""
    if not tasks:
        print("No tasks to complete.")
        return

    print("\n--- Tasks to Complete ---")
    for i, task in enumerate(tasks):
        priority_map = {1: 'High', 2: 'Medium', 3: 'Low'}
        print(f"{i+1}. Name: {task['name']}, Priority: {priority_map[task['priority']]}")
    print("-------------------------\n")

    while True:
        try:
            task_index_input = input("Enter the number of the task to complete (or 'c' to cancel): ").strip()
            if task_index_input.lower() == 'c':
                print("Task completion cancelled.")
                return

            task_index = int(task_index_input) - 1 # Adjust for 0-based indexing

            if 0 <= task_index < len(tasks):
                completed_task = tasks.pop(task_index) # Remove from list
                task_name = completed_task['name']
                now = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")

                with open('completed_tasks.txt', 'a') as f:
                    f.write(f"{now} - {task_name}\n")
                print(f"Task '{task_name}' marked as complete and saved to 'completed_tasks.txt'.")
                break
            else:
                print("Invalid task number. Please enter a number from the list.")
        except ValueError:
            print("Invalid input. Please enter a number or 'c'.")

def main():
    """Main function to run the task manager."""
    while True:
        print("\nPersonal Task Manager Menu:")
        print("1. Add a task")
        print("2. View tasks")
        print("3. Complete a task")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ")

        if choice == '1':
            add_task()
        elif choice == '2':
            view_tasks()
        elif choice == '3':
            complete_task()
        elif choice == '4':
            print("Exiting Task Manager. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter 1, 2, 3, or 4.")

# Run the main program
if __name__ == "__main__":
    main()
```
